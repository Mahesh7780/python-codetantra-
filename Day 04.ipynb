```python
t=1234
while t>=10:
    s=0
    while t>0:
        s+=t%10  #4+3+2+1=10  2--> 0 + 1 = 1 loop stops (while is False 1<10)
        t=t//10 #123,12,1     2--> 10,1
    t=s # t=10  while condition is true next loop (now t=1)
print(t) #printing t
```

Armstrong Number

```python
t=123
tt=t
l=len(str(t))
s=0
while t>0:
    c=t%10
    s=s+c**l
    t=t//10
if tt==s:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

Strong Number

```python
t=145
f=0
r=t
while t>0:  #true condition ,  14>0 ,1>0 ,contion False
    m=t%10 #last element , 5 , 1
    s=1
    for i in range(1,m+1): # 1 to 5+1(len) , #1 to 4+1 , 1
        s=s*i  # 1*1,1*2,2*3,6*4,24*5 ==>120 , 1*2*2*3*6*4==>24 , 1==>1
    f=f+s # f=120, f=120+24 ,f=144+1
    t=t//10 #14 #1 #0

if r==f:
    print("Strong Number")
else:
    print("Not Strong Number")
```

Spy Number

```python
t=1124
r=1
m=0
while t>0:
    c=t%10 #last num
    r=r*c # mul
    m=m+c # add
    t=t//10 # removing last num
if r==m:
    print("Spy Number")
else:print("Not Spy Number")
```

Magic Number

```python
n=23
while n > 9:
    m = 0

    while n > 0:
        c = n % 10
        m += c
        n = n // 10

    n = m

if n == 1:
    print("Magic Number")
else:
    print("Not Magic Number")
```
