Prime Number Spiral Matrix

```python
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True


def generate_primes(n):
    primes = []
    num = 2

    while len(primes) < n:
        if is_prime(num):
            primes.append(num)
        num += 1

    return primes


n = 5
matrix = [[0] * n for _ in range(n)]

primes = generate_primes(n * n)

top = 0
bottom = n - 1
left = 0
right = n - 1

index = 0

while top <= bottom and left <= right:

    # Left to Right
    for i in range(left, right + 1):
        matrix[top][i] = primes[index]
        index += 1
    top += 1

    # Top to Bottom
    for i in range(top, bottom + 1):
        matrix[i][right] = primes[index]
        index += 1
    right -= 1

    # Right to Left
    for i in range(right, left - 1, -1):
        matrix[bottom][i] = primes[index]
        index += 1
    bottom -= 1

    # Bottom to Top
    for i in range(bottom, top - 1, -1):
        matrix[i][left] = primes[index]
        index += 1
    left += 1


for row in matrix:
    print(row)
```
