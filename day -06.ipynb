# Python Functions

Functions are the fundamental building blocks of reusable, modular, and maintainable code in Python. This guide covers absolutely everything you need to know about Python functions, from basic syntax to advanced functional programming concepts.

## 1. Introduction to Functions

### What is a Function?
A function is a named block of reusable code designed to perform a specific task. You define it once, and you can "call" or execute it as many times as needed.

### Why Use Functions?
*   **Reusability (DRY - Don't Repeat Yourself):** Avoid duplicating code.
*   **Modularity:** Break down complex problems into smaller, manageable pieces.
*   **Readability:** Descriptive function names make your code easier to read.
*   **Maintainability:** Fixing a bug in a function fixes it everywhere the function is used.
*   **Testing:** Smaller, isolated pieces of code are easier to test.

## 2. Defining and Calling Functions

### Basic Syntax
In Python, functions are defined using the `def` keyword.

```python
def greet(name):
    """This is a docstring. It explains what the function does."""
    print(f"Hello, {name}!")

# Calling the function
greet("Alice")  # Output: Hello, Alice!
```

**Key Components:**
1.  **`def` keyword:** Signals the start of a function definition.
2.  **Function Name:** Follows `snake_case` naming conventions.
3.  **Parentheses `()`:** Contain parameters (optional).
4.  **Colon `:`:** Marks the beginning of the function body.
5.  **Indentation:** The body of the function MUST be indented (standard is 4 spaces).

### The `pass` Statement
If you are designing a function but haven't written the body yet, use `pass` to avoid a syntax error.

```python
def future_function():
    pass # To be implemented later
```

## 3. Parameters and Arguments

*   **Parameter:** The variable defined in the function signature.
*   **Argument:** The actual value passed to the function when it is called.

### 3.1. Positional Arguments
Arguments are mapped to parameters based on their position.

```python
def subtract(a, b):
    return a - b

print(subtract(10, 5)) # 10 maps to a, 5 maps to b. Output: 5
```

### 3.2. Keyword Arguments
You can specify the parameter name when calling the function, allowing you to ignore the order.

```python
print(subtract(b=5, a=10)) # Output: 5
```

### 3.3. Default Arguments
You can assign default values to parameters. If an argument is not passed, the default is used. **Rule:** Default parameters must come *after* non-default parameters.

```python
def greet(name, msg="Good morning!"):
    print(f"Hello {name}, {msg}")

greet("Bob")                  # Output: Hello Bob, Good morning!
greet("Alice", "How are you?") # Output: Hello Alice, How are you?
```

> [!WARNING]
> **The Mutable Default Argument Gotcha**
> Never use mutable objects (like lists or dictionaries) as default arguments. They are evaluated only *once* when the function is defined, leading to unexpected behavior.

```python
# BAD
def append_to_list(val, my_list=[]):
    my_list.append(val)
    return my_list

print(append_to_list(1)) # [1]
print(append_to_list(2)) # [1, 2] - Wait, where did 1 come from?

# GOOD
def append_to_list_safe(val, my_list=None):
    if my_list is None:
        my_list = []
    my_list.append(val)
    return my_list
```

### 3.4. Variable-Length Arguments

#### `*args` (Non-Keyword Arguments)
Allows a function to accept any number of positional arguments as a **tuple**.

```python
def sum_all(*args):
    total = 0
    for num in args:
        total += num
    return total

print(sum_all(1, 2, 3, 4)) # Output: 10
```

#### `**kwargs` (Keyword Arguments)
Allows a function to accept any number of keyword arguments as a **dictionary**.

```python
def print_user_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_user_info(name="Alice", age=30, job="Engineer")
```

### 3.5. Argument Unpacking
You can unpack a list/tuple or dictionary when *calling* a function.

```python
def multiply(a, b, c):
    return a * b * c

nums = [2, 3, 4]
print(multiply(*nums)) # Unpacks list to positional args. Output: 24

data = {'a': 2, 'b': 3, 'c': 4}
print(multiply(**data)) # Unpacks dict to keyword args. Output: 24
```

### 3.6. Positional-Only and Keyword-Only Arguments
Python 3.8 introduced advanced control over how arguments are passed.

*   **`/`**: Parameters *before* the slash must be positional-only.
*   **`*`**: Parameters *after* the asterisk must be keyword-only.

```python
def complex_func(pos_only, /, standard, *, kw_only):
    print(pos_only, standard, kw_only)

# complex_func(1, 2, 3) # ERROR: kw_only must be keyword
# complex_func(pos_only=1, standard=2, kw_only=3) # ERROR: pos_only cannot be keyword
complex_func(1, standard=2, kw_only=3) # VALID
complex_func(1, 2, kw_only=3)          # VALID
```

## 4. Return Values

Functions use the `return` statement to send data back to the caller.

### Returning Single and Multiple Values
If a function doesn't have a `return` statement, it implicitly returns `None`. Python can return multiple values by automatically packing them into a tuple.

```python
def get_user():
    name = "Alice"
    age = 30
    return name, age # Returns a tuple: ("Alice", 30)

user_name, user_age = get_user() # Tuple unpacking
```

### Early Return
`return` immediately exits the function. It is often used for "guard clauses".

```python
def divide(a, b):
    if b == 0:
        return "Cannot divide by zero!" # Early return
    return a / b
```

## 5. Scope and Namespaces (The LEGB Rule)

Python resolves variables in a specific order: **L**ocal, **E**nclosing, **G**lobal, **B**uilt-in.

1.  **Local:** Variables defined inside the current function.
2.  **Enclosing:** Variables defined in the outer (calling) function (for nested functions).
3.  **Global:** Variables defined at the top level of the script.
4.  **Built-in:** Python's pre-defined names (e.g., `len()`, `print()`).

### The `global` Keyword
Used to modify a global variable from inside a local scope.

```python
counter = 0

def increment():
    global counter
    counter += 1

increment()
print(counter) # Output: 1
```

### The `nonlocal` Keyword
Used in nested functions to modify a variable in the enclosing (non-global) scope.

```python
def outer():
    x = "local"
    def inner():
        nonlocal x
        x = "nonlocal"
        print("Inner:", x)
    inner()
    print("Outer:", x)

outer()
# Output:
# Inner: nonlocal
# Outer: nonlocal
```

## 6. Advanced Function Types & Concepts

### 6.1. First-Class Functions
In Python, functions are "first-class citizens." They can be:
*   Assigned to variables.
*   Passed as arguments to other functions.
*   Returned from other functions.

```python
def shout(text):
    return text.upper()

yell = shout # Assign to variable
print(yell("hello")) # Output: HELLO
```

### 6.2. Higher-Order Functions
Functions that take other functions as arguments or return them. Built-in examples include `map`, `filter`, and `reduce`.

```python
def apply_operation(func, x, y):
    return func(x, y)

def add(a, b): return a + b

print(apply_operation(add, 5, 3)) # Output: 8
```

### 6.3. Lambda Functions (Anonymous Functions)
Small, single-expression functions without a name. Often used for short callbacks.

**Syntax:** `lambda arguments: expression`

```python
square = lambda x: x ** 2
print(square(5)) # Output: 25

# Using with sorting
points = [(1, 2), (3, 1), (5, -1)]
points.sort(key=lambda point: point[1]) # Sort by the second element (Y-coordinate)
print(points) # Output: [(5, -1), (3, 1), (1, 2)]
```

### 6.4. Generator Functions (`yield`)
Instead of computing all values at once and returning them, generators `yield` one value at a time, pausing their state. This is highly memory-efficient for large datasets.

```python
def count_up_to(max):
    count = 1
    while count <= max:
        yield count
        count += 1

counter = count_up_to(3)
print(next(counter)) # Output: 1
print(next(counter)) # Output: 2
```

### 6.5. Recursive Functions
Functions that call themselves to solve a smaller instance of the same problem. They MUST have a base case to avoid infinite recursion.

```python
def factorial(n):
    if n == 1 or n == 0: # Base case
        return 1
    return n * factorial(n - 1) # Recursive call
```

## 7. Decorators

Decorators are a powerful way to modify the behavior of a function without changing its source code. They are essentially wrappers.

### Creating a Decorator

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_whee():
    print("Whee!")

say_whee()
```

### Decorating Functions with Arguments
To decorate functions that take arguments, use `*args` and `**kwargs` in the wrapper.

```python
import time
from functools import wraps

def timer_decorator(func):
    @wraps(func) # Preserves original function's name and docstring
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Function {func.__name__} took {end_time - start_time:.4f} seconds")
        return result
    return wrapper

@timer_decorator
def slow_add(a, b):
    time.sleep(1)
    return a + b

slow_add(5, 7)
```

## 8. Asynchronous Functions (`async` / `await`)

Used for concurrent programming (I/O bound tasks like network requests). Defined with `async def`. They return a coroutine object which must be `await`ed.

```python
import asyncio

async def fetch_data():
    print("Start fetching...")
    await asyncio.sleep(2) # Simulates network request
    print("Done fetching!")
    return {"data": 123}

async def main():
    result = await fetch_data()
    print(result)

# To run: asyncio.run(main())
```

## 9. Documentation and Type Hinting

Modern Python heavily relies on explicit documentation and type hinting for large codebases.

### Type Hints (PEP 484)
Specify expected input and output types. (Note: Python doesn't enforce these at runtime; they are for IDEs and tools like `mypy`).

```python
def format_greeting(name: str, age: int) -> str:
    return f"{name} is {age} years old."
```

### Docstrings (PEP 257)
Always document what your function does, its arguments, and its return values. The Google standard is highly recommended.

```python
def fetch_user_data(user_id: int) -> dict:
    """
    Fetches user data from the database.

    Args:
        user_id (int): The unique identifier for the user.

    Returns:
        dict: A dictionary containing 'name', 'email', and 'role'.

    Raises:
        ValueError: If user_id is negative.
    """
    if user_id < 0:
        raise ValueError("User ID cannot be negative")
    return {"name": "Alice", "email": "alice@email.com", "role": "admin"}
```

## 10. Best Practices (Clean Code)

1.  **Single Responsibility Principle:** A function should do *one* thing and do it well. If it has "and" in its name (e.g., `calculate_and_print`), split it into two functions.
2.  **Naming:** Use descriptive verbs (`get_user()`, `calculate_total()`).
3.  **Avoid Global Variables:** Rely on arguments and return values. This makes functions "pure" and predictable.
4.  **Keep it short:** Ideally, a function should fit on a single screen.
5.  **Fail Fast:** Handle edge cases and raise exceptions at the *beginning* of the function using early returns or assertions.
